```python
total_customers = len(df)

churned_customers = (df["churn"] == "Yes").sum()

churn_rate = (churned_customers / total_customers) * 100

retained_customers = (df["churn"] == "No").sum()

retention_rate = (retained_customers / total_customers) * 100

average_tenure = df["tenure_months"].mean()

average_monthly_charges = df["monthly_charges"].mean()

average_total_charges = df["total_charges"].mean()

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Retention Rate:", round(retention_rate, 2), "%")
print("Average Tenure:", round(average_tenure, 2), "months")
print("Average Monthly Charges:", round(average_monthly_charges, 2))
print("Average Total Charges:", round(average_total_charges, 2))
```
